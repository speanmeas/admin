```python
# Convert image to favicon
import cv2
import numpy as np

# Read the source image
img = cv2.imread('../asset/logo.png')

# Convert to favicon size (32x32 pixels)
favicon_size = (32, 32)
favicon = cv2.resize(img, favicon_size, interpolation=cv2.INTER_AREA)

# Save as favicon.png
cv2.imwrite('../asset/favicon.png', favicon)

print(f"Favicon created and saved as ../asset/favicon.png")
print(f"Original size: {img.shape[:2]}")
print(f"Favicon size: {favicon.shape[:2]}")

# Display the favicon
from matplotlib import pyplot as plt
plt.imshow(cv2.cvtColor(favicon, cv2.COLOR_BGR2RGB))
plt.title('Favicon (32x32)')
plt.axis('off')
plt.show()
```

```python
# Convert image to icon-192.png and icon-512.png
import cv2
import numpy as np

# Read the source image
img = cv2.imread('../asset/logo.png')

# Convert to icon-192 size (192x192 pixels)
icon_192_size = (192, 192)
icon_192 = cv2.resize(img, icon_192_size, interpolation=cv2.INTER_AREA)

# Convert to icon-512 size (512x512 pixels)
icon_512_size = (512, 512)
icon_512 = cv2.resize(img, icon_512_size, interpolation=cv2.INTER_AREA)

# Save the icons
cv2.imwrite('../asset/icon-192.png', icon_192)
cv2.imwrite('../asset/icon-512.png', icon_512)

print(f"Icons created and saved:")
print(f"icon-192.png: {icon_192.shape[:2]} pixels")
print(f"icon-512.png: {icon_512.shape[:2]} pixels")
print(f"Original size: {img.shape[:2]}")

# Display both icons
from matplotlib import pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

# Display icon-192
ax1.imshow(cv2.cvtColor(icon_192, cv2.COLOR_BGR2RGB))
ax1.set_title('Icon-192 (192x192)')
ax1.axis('off')

# Display icon-512
ax2.imshow(cv2.cvtColor(icon_512, cv2.COLOR_BGR2RGB))
ax2.set_title('Icon-512 (512x512)')
ax2.axis('off')

plt.tight_layout()
plt.show()
```
